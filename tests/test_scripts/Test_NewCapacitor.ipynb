This code demonstrates the "NewCapacitor" function written for Object Builder.

Step1: Initialization and loading network from XML file

```python
import os
from cimgraph.databases import XMLFile
from cimgraph.models import FeederModel
import cimgraph.data_profile.cimhub_2023 as cim
os.environ['CIMG_CIM_PROFILE'] = 'cimhub_2023'
file = XMLFile(filename='../test_models/IEEE13.xml')
feeder = cim.Feeder(mRID= '49AD8E07-3BF9-A4E2-CB8F-C3722F837B62')
network = FeederModel(container=feeder, connection=file)
```

Step2: Loading the Object Builder Library as Builder (this can be further used to call the functions of individual objects under OB).

```python
import cimbuilder.object_builder as builder
```

Step3: providing inputs for the new equipment. Here the equipment is Capacitor. So only one node required to connect. In IEEE 13 node network, node 1 = 646 is selected to connect capacitor.

Other inputs for synchronous generator are:
* sectionNumber
* b
* b0
* g
* g0





```python

node_capa_1 = '646'

sectionNumber = 1
b = 0.5
b0 = 0.1
g = 1
g0 = 0.1
```

Step4: Call the OB function using the inputs provided.

```python
name = 'capacitor_1'
Capacitor_646 = builder.new_capacitor(network, feeder, name, node_capa_1, sectionNumber, b, b0, g, g0)  
```

Step5: Verifying that the new object was created
