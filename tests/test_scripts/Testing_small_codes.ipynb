1. Create a new dictionary B by modifying A. Print entire B, but highlight newer parts of B compared to A.

```python
def highlight_differences(dictA, dictB):
    # ANSI escape codes for coloring text
    blue_start = '\033[34m'
    red_bold_start = '\033[1;31m'
    reset_style = '\033[0m'
    
    for key, value in dictB.items():
        # Check if the key is present in A with the same value
        if key not in dictA or dictB[key] != dictA[key]:
            # Print in bold red if the part is different
            print(f"{red_bold_start}Key: {key}, Value: {value}{reset_style}")
        else:
            # Print in blue if the part is the same
            print(f"{blue_start}Key: {key}, Value: {value}{reset_style}")

# Example dictionaries
dictA = {'key1': 'value1', 'key2': 'value2', 'key3': 'value3'}
dictB = {'key2': 'value2', 'key3': 'different_value3', 'key4': 'value4'}

# Call the function
highlight_differences(dictA, dictB)
```

The code block above prints the normal portions in blue and the highlighted portions in red. This will find application in showing the changes made to the XML file while using Object Builder. The original part of the XML file cna be shown in blue, and newly added component after using OB can displayed in red. 

2. Attempting to save xml file as csv. I think this might give us better way to look at xml giles?

```python
import xml.etree.ElementTree as ET
import csv

def xml_to_csv(xml_file, csv_file):
    # Parse the XML file
    tree = ET.parse(xml_file)
    root = tree.getroot()

    # Open a CSV file for writing
    with open(csv_file, 'w', newline='') as csvfile:
        csvwriter = csv.writer(csvfile)
        
        # Extract column names (headers) from the first element
        headers = [elem.tag for elem in root[0]]
        csvwriter.writerow(headers)  # Write headers to the CSV

        # Iterate through each entry in the XML and write it to the CSV
        for entry in root:
            row = [subelem.text for subelem in entry]
            csvwriter.writerow(row)

# Usage
xml_file = '../test_models/IEEE13.xml'
csv_file = 'IEEE13_XML_as_CSV.csv'
xml_to_csv(xml_file, csv_file)
```
