This code demonstrates the "NewTransmissionLine" function written for Object Builder.

Step1: Initialization and loading network from XML file

```python
import os
from cimgraph.databases import XMLFile
from cimgraph.models import FeederModel
import cimgraph.data_profile.cimhub_2023 as cim
os.environ['CIMG_CIM_PROFILE'] = 'cimhub_2023'
file = XMLFile(filename='../test_models/IEEE13.xml')
feeder = cim.Feeder(mRID= '49AD8E07-3BF9-A4E2-CB8F-C3722F837B62')
network = FeederModel(container=feeder, connection=file)
```

Step2: Loading the Object Builder Library as Builder (this can be further used to call the functions of individual objects under OB).

```python
import cimbuilder.object_builder as builder
```

Step3: providing inputs for the new equipment. Here the equipment is Transmission Line. So Two nodes are required to connect.

In IEEE 13 node network, node 1 = 646 is selected as the form node to connect the TL. For the "to node" a new node is created. 

Other inputs for synchronous generator are:
* Length
* r
* x
* bch
* r0
* x0
* bch0





```python

node_TL_1 = '646'

node_TL_2 = '645' ## a new node can be created for the new TL

sectionNumber = 1
length = 6.5
r = 0.002
x = 0.03
bch = 0.5
r0 = 0.0002
x0 = 0.003
bch0 = 0.05
```

Step4: Call the OB function using the inputs provided.

```python
name = 'TL_645_946_new'
TL_645_646 = builder.new_transmission_line(network, feeder, name, node_TL_1 , node_TL_2, length, r, x, bch, r0, x0, bch0)  
```

Step5: Verifying that the new object was created

```python
TL_645_646.__dict__ ## extracting breaker attributes in dictionary form
```
