Switch (vaccum recloser switch) parameter info saved to cim format from OEM data sheets

```python
## Importing neccessaruy libraries
import cimgraph.data_profile.cim17v40 as cim
import os
import cimgraph.data_profile.cim17v40 as cim
from cimgraph.models import FeederModel
from cimgraph.databases import XMLFile
from cimgraph import utils
from mermaid import Mermaid
import numpy as np
import pandas as pd
### Initialize cim profile and an empty XML file
os.environ['CIMG_CIM_PROFILE'] = 'cimgraph.data_profile.cim17v40'
namespaces = {'cim':'http://iec.ch/TC57/CIM100#',
              'gad':'http://gridappsd.org/CIM/extension#'}
file = XMLFile(filename='', namespaces=namespaces)
network = FeederModel(container=None, connection=file)
```

```python


df_Siemens_Switch_para_table = pd.read_csv('Siemens_Vaccum_recloser_CIM_data.csv')
## Link for Siemens recloser data set : https://assets.new.siemens.com/siemens/assets/api/uuid:8259a7dcc82300bcaff9ae63755dccb57091c758/catalog-hg-11-42-recloser-en.pdf

SW_rated_voltage_kV = df_Siemens_Switch_para_table.values[:,0]
#print(SW_rated_voltage_kV )
SW_rated_Impulse_Withstand_Volt_kV = df_Siemens_Switch_para_table.values[:,1]
SW_Short_Duration_withstand_voltage_kV = df_Siemens_Switch_para_table.values[:,2]
SW_breaking_capacity_kA = df_Siemens_Switch_para_table.values[:,3]
SW_rated_current_A = df_Siemens_Switch_para_table.values[:,4]
SW_1phase_3phase = df_Siemens_Switch_para_table.values[:,5]

for k in range(df_Siemens_Switch_para_table.shape[0]):
    #print(k)

    recloser_info = cim.SwitchInfo(name='Siemens_3AD3126_'+str(k))
    recloser_info.description = 'Siemens vacuum Recloser 3AD - ' + str(SW_rated_current_A[k]) + 'A'
    recloser_info.ratedVoltage = cim.Voltage(SW_rated_voltage_kV[k], input_unit='kV')
    recloser_info.ratedImpulseWithstandVoltage = cim.Voltage(SW_rated_Impulse_Withstand_Volt_kV[k], input_unit='kV')
    # # recloser_info.ratedShortDurationVoltage = SW_Short_Duration_withstand_voltage_kV[k] # not supported by CIM yet
    recloser_info.breakingCapacity = cim.CurrentFlow(SW_breaking_capacity_kA[k], input_unit = 'kA')
    recloser_info.ratedCurrent = cim.CurrentFlow(SW_rated_current_A[k], input_unit = 'A')
    # # recloser_info.phaseCount = cim.PhaseCountKind.threePhase # CIM18 attribute

    if(SW_1phase_3phase[k] == 1):
        recloser_info.isSinglePhase = True
    else:
        recloser_info.isSinglePhase = False

    network.add_to_graph( recloser_info)
    # # recloser_info.isUnganged = True
    # recloser_equipment = cim.Recloser(name='recloser_' + str(k))
    # recloser_equipment.AssetDatasheet = recloser_info
    # recloser_equipment.ratedCurrent = recloser_info.ratedCurrent
    # recloser_equipment.breakingCapacity = recloser_info.breakingCapacity


```
