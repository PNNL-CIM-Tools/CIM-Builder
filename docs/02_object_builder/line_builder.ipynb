```python
import os
os.environ['CIMG_CIM_PROFILE'] = 'cim18v17'
os.environ['CIMG_USE_UNITS'] = "TRUE"
os.environ['CIMG_SYSTEM_MVA_BASE'] = '100'
from cimtables import Qty
import cimprofiles.cim18.cim18v17 as cim
from cimgraph.models import FeederModel
new_feeder = cim.Feeder(name='demo_test_feeder')
network = FeederModel(container=new_feeder, connection=None)
```

```python
base_volt = cim.BaseVoltage(name='bv_12.47kV', 
            nominalVoltage=cim.Voltage(12.47,'kV'))
network.add_to_graph(base_volt)
```

```python
from cimbuilder.builders.line import ACLineSegmentBuilder
line_builder=ACLineSegmentBuilder(network=network, container=new_feeder)
```

```python
line1 = line_builder.create(name = 'line1')
line_builder.add_connectivity(line=line1, node1='bus1', node2='bus2',
    length=Qty(5,'mile'), circuitNumber=1, BaseVoltage=base_volt.mRID)
line_builder.add_electrical(line=line1, r=Qty(0.05, 'pu'), x=Qty(0.10, 'pu'))
```
