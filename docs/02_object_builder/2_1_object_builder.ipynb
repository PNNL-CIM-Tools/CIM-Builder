# Object Builder Overview

The `object_builder` module provides shortcut functions for building sub-graphs of CIM objects as needed for each new datatype. 

For example, when creating a single-phase line, it is necessary to create the ACLineSegment, ACLineSegmentPhase, two Terminal objects, and associate all of the objects with each other. CIM-Builder provides that functionality in a set of simple wrapper scripts that chain the creation of all required objects, attributes, and associations.

The object_builder module can be imported by running

```python
from cimbuilder import object_builder
```

The object_builder module offers the following methods:

* `new_one_terminal_object()`: Generic method for `ConductingEquipment` classes that require one `Terminal`, such as capacitors and loads

* `new_two_terminal_object()`: Generic method for `ConductingEquipment` classes that require two `Terminal` objects, such as lines and switches
