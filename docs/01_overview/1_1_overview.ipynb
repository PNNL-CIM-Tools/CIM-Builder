# Overview

CIM-Builder is a Python library from Pacific Northwest National Laboratory for building CIM power system
models from scratch, with no model file to start from. Other CIM tooling converts an existing source, such
as an OpenDSS, PSS/E or GIS model; CIM-Builder writes the CIM objects directly, through one call per piece
of equipment or per substation. It can:

1. create new node-breaker substations from interactive API calls,
2. insert existing CIM distribution feeders into those substations,
3. add new equipment and measurements to an existing model, and
4. insert aggregate feeder data into an existing CIM transmission model.

## Where it fits

CIM-Builder sits on top of the other PNNL CIM tools. [cim-graph](https://pnnl-cim-tools.github.io/CIM-Graph/)
holds the model in memory and reads and writes CIM XML and JSON-LD, and
[cimtables](https://pnnl-cim-tools.github.io/CIM-Tables/) does the attribute writing, unit conversion and
association resolution that every builder calls into.

```mermaid
flowchart LR
    profiles["<b>cim-profiles</b><br/>CIM classes and units"] --- cimgraph["<b>cim-graph</b><br/>the model in memory,<br/>CIM XML and JSON-LD"]
    cimgraph --- tables["<b>cimtables</b><br/>attribute and<br/>association writes"]
    tables --- builder["<b>CIM-Builder</b><br/>object and<br/>substation builders"]
```

## What is in the package

| Module | What it builds |
|---|---|
| [`cimbuilder.object_builder`](../03_object_builder/3_1_object_builders.ipynb) | One builder class per CIM class (`ACLineSegmentBuilder`, `BreakerBuilder`, `EnergyConsumerBuilder`, ...). Each one creates the object, then fills in one CIM profile at a time: connectivity, electrical parameters, short-circuit data, steady-state hypothesis and so on. |
| [`cimbuilder.substation_builder`](../02_substation_builder/2_1_substation_builder.ipynb) | One builder per node-breaker substation layout: single bus, sectionalized bus, main and transfer, double bus single breaker, ring bus and breaker and a half. Each builds the buses and every breaker, disconnector and junction node of the layout. |
| `cimbuilder.feeder_builder` | `new_aggregate_feeder`, an aggregate feeder for a transmission model. |

Builders write into a cim-graph model, such as a `FeederModel` loaded from CIM XML, and refer to other
objects by mRID string. The [substation builder](../02_substation_builder/2_1_substation_builder.ipynb)
pages walk through each layout, inserting the IEEE 13 node test feeders into a new substation, and
the [object builder](../03_object_builder/3_1_object_builders.ipynb) pages through each kind of equipment.
