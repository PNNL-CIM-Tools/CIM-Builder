# Overview

CIM-Builder is a new python library developed by Pacific Northwest National Laboratory for creating CIM models "from scratch" with no pre-existing model files. This is a significantly different capability than all other tooling, which requires a source file from which to start, such as OpenDSS, PSSE, or GIS data. The library currently inlcudes three main functionalities:

1) automatic creation of new node-breaker substations in CIM based on interactive API calls

2) automatic insertion of existing distribution feeders in CIM into new node-breaker substations

3) automatic insertion of new objects and measurements into existing models

4) automatic insertion of new aggregate feeder data in existing CIM transmission models





## Substation Builder

The `substation_builder` module has one builder class per common node-breaker substation layout:

* `SingleBusSubstationBuilder`
* `SectionalizedBusSubstationBuilder`
* `MainAndTransferSubstationBuilder`
* `DoubleBusSingleBreakerSubstationBuilder`
* `RingBusSubstationBuilder`
* `BreakerAndHalfSubstationBuilder`

They follow the same pattern as the `object_builder` classes. Each one is constructed with the CIMantic Graphs network and container it writes into. Every method takes the object it extends plus keyword arguments, with references given as mRID strings. The methods are called in order:

1. `create()` makes the `Substation`.
2. `add_voltage_level(substation, BaseVoltage=...)` adds one `VoltageLevel` per nominal voltage.
3. `add_buses(voltage_level, ...)` adds the busbars and bus ties of the layout.
4. `add_branch(voltage_level, ..., node=...)` adds the breaker, disconnector and junction chain between a bus and a branch node. The node can be a new node for a transmission line, transformer or shunt, or the `sourcebus` of an existing distribution feeder.

To insert a distribution feeder, load it into the same graph as the substation, call `add_branch` with the feeder's `sourcebus` node, and link the feeder to the substation with `FeederBuilder.add_connectivity(feeder, NormalEnergizingSubstation=...)`. The feeder's head node keeps its Feeder container. The substation's own nodes and switches go in its `VoltageLevel`.
