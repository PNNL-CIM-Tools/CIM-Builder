# Installation

CIM-Builder needs Python 3.10 or newer.

## From PyPI

```bash
pip install --pre cim-builder
```

`cim-builder` is imported as `cimbuilder`. `--pre` is required for now: CIM-Builder is published as alpha
releases and depends on pre-release versions of [cim-tables](https://pnnl-cim-tools.github.io/CIM-Tables/),
[cim-graph](https://pnnl-cim-tools.github.io/CIM-Graph/) and
[cim-profiles](https://github.com/PNNL-CIM-Tools/CIM-Profiles).

With [uv](https://docs.astral.sh/uv/):

```bash
uv add --prerelease allow cim-builder
```

## From source

Clone the repository and let uv create the virtual environment in `.venv` and install CIM-Builder with
its dependencies:

```bash
git clone https://github.com/PNNL-CIM-Tools/CIM-Builder.git -b develop
cd CIM-Builder
uv sync
uv run pytest
```

To use your checkout from another project's environment, install it there as an editable package:

```bash
pip install -e ../CIM-Builder
```
