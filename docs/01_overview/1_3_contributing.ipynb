# Contribution Guidelines

We welcome contributions to CIM-Builder. These guidelines keep contributions easy to review and accept.

## Getting started

You need a GitHub account, git, and [uv](https://docs.astral.sh/uv/) to manage the Python environment.

## How to contribute

### Fork and clone the repository

1. On GitHub, click **Fork** on the
   [CIM-Builder repository](https://github.com/PNNL-CIM-Tools/CIM-Builder) to copy it to your account.
2. Clone your fork and install the development environment:
    ```bash
    git clone https://github.com/YOUR_USERNAME/CIM-Builder.git
    cd CIM-Builder
    uv sync
    ```
3. Create a branch for your changes, starting from `develop`:
    ```bash
    git checkout develop
    git checkout -b feature/your-feature-name
    ```

### Make and test your changes

Follow the [coding standards](#coding-standards) below, and run the tests before you commit:

```bash
uv run pytest
```

If you change a documented behaviour, re-run the notebooks that show it (see
[Building these docs](../developer/index.md#building-these-docs)).

### Commit and push

All commits must be signed. You can configure commit signing by following
[GitHub's instructions](https://docs.github.com/en/authentication/managing-commit-signature-verification/signing-commits).

```bash
git add .
git commit -S -m "Add feature: description of your feature"
git push origin feature/your-feature-name
```

### Open a pull request

Open a pull request from your branch against `develop`, with a clear title and a description of what it
changes and why.

## Coding standards

1. **PEP 8**: follow the [PEP 8](https://peps.python.org/pep-0008/) style guide, and match the style of
   the surrounding code.
2. **Docstrings**: give every builder class and method a docstring that lists its keyword arguments.
3. **Type annotations**: annotate function arguments and return values.
4. **Tests**: add unit tests for new builders and logic, under `tests/`, and make sure all tests pass.

## Review process

1. A maintainer reviews each pull request.
2. Review comments may ask for changes. Push them to the same branch to update the pull request, with
   signed commits.
3. Once approved, the pull request is merged into `develop`.

Thank you for helping to improve CIM-Builder.
