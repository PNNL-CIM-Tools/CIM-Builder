The objective of OB is to expand SB to include all realistic elements in power systems. A core feature of the OB, similar to substation builder will be to create elements from scratch and using CIM and connect them realistically. While the end goal as of now is creating an entire power system network from scratch, the goal for this FY is limited to read a network using XML and connect new elements to it makign it a larger network.

```python
import os
from cimgraph.databases import XMLFile
from cimgraph.models import FeederModel
import cimgraph.data_profile.cimhub_2023 as cim
os.environ['CIMG_CIM_PROFILE'] = 'cimhub_2023'
file = XMLFile(filename='../tests/test_models/IEEE13.xml')
feeder = cim.Feeder(mRID= '49AD8E07-3BF9-A4E2-CB8F-C3722F837B62')
network = FeederModel(container=feeder, connection=file)
 
```
