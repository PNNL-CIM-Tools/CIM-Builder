```python
import cimgraph.data_profile.cim17v40 as cim

import os
import cimgraph.data_profile.cim17v40 as cim
from cimgraph.models import FeederModel
from cimgraph.databases import XMLFile
from cimgraph import utils
from mermaid import Mermaid
 
 
os.environ['CIMG_CIM_PROFILE'] = 'cimgraph.data_profile.cim17v40'
 
 
namespaces = {'cim':'http://iec.ch/TC57/CIM100#',
              'gad':'http://gridappsd.org/CIM/extension#'}
 
file = XMLFile(filename='', namespaces=namespaces)
network = FeederModel(container=None, connection=file)
```

Poles data sheet: https://www.rd.usda.gov/media/file/download/uep-bulletin-1728f-804.pdf

Pole parameters are
1. height
2. spacing
3. position

4. xcoord
5. ycoord


#########################

```python


template = cim.WireAssemblyInfo(name='USDA_A1.6NP')
pole = cim.Pole(name='USDA_A1.6NP_pole_info')
pole.height = cim.Length(40, input_unit='ft')
spacing = cim.WireSpacingInfo(name='USDA_A1.6NP_spacing')
position_phs = cim.WirePosition(name='USDA_A1.6NP_phs_pos')
position_phs.sequenceNumber = 1
position_phs.xCoord = cim.Displacement(-44, input_unit='in')
position_phs.yCoord = pole.height
position_neut = cim.WirePosition(name='USDA_A1.6NP_neutral_pos')
position_neut.sequenceNumber = 2
position_neut.xCoord = cim.Displacement(44, input_unit='in')
position_neut.yCoord = pole.height.quantity - cim.Length(18, input_unit='in').quantity ### use this format for all conductors with different height

position_phs.WireSpacingInfo = spacing
position_neut.WireSpacingInfo = spacing
template.WireSpacingInfo = spacing

network.add_to_graph(template)
network.add_to_graph(pole)
network.add_to_graph(spacing)
network.add_to_graph(position_phs)
network.add_to_graph(position_neut)

utils.write_xml(network, 'pole.xml', namespaces)
 
```
