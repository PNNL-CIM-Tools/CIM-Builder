The asset parameters from wires dat sheets of various OEMs are collected and saved here

OEMs for Wires
1. Southwire
2. Prysmian
3. Nexans
4. Sterlite electric


For both Southwire and Prysmian, the datasheets for following types of transmission lines are available

1. ACSR (Aluminium Conductor Steel Reinforces) Bare
2. ACSR TW (Trapezoidal Shaped)
3. ACSS (Aluminium Conductor Steel Supported)
4. Ultra high strength ACSS

Links for multiple types can be acccessed from following links

1. Southwire A - https://www.southwire.com/wire-cable/bare-aluminum-overhead-transmission-distribution/c/c-bare-aluminum-overhead
2. Prysmian UK data sheets - https://na.prysmian.com/markets/power-grid/overhead-transmission
3. Nexans - https://www.nexans.us/en/Product-Datasheets/Utility-Transmission-Distribution/Bare-Overhead-Conductors.html



Data available are

1. Southwire - Size, stranding, Diameter, Weights, Content, Rated strength (lbs), Resistance, Ampacity

2. Prysmian - Size, stranding, diameter, mass, rated streength, resistance, reactance, amapacity


3a. Nexans Physical data - Code word, Size, No of wires, Diameter for diff types, Weights, rated strength (ldf)


3b. Code word, Size - Resistance and Reactance at different temperatures, Ampacity
